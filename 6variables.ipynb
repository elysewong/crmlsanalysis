```python
Q1_LivingArea = df.get('LivingArea').quantile(0.25)
Q1_LivingArea
```

```python
Q3_LivingArea = df.get('LivingArea').quantile(0.75)
Q3_LivingArea
```

```python
IQR_LivingArea = Q3_LivingArea - Q1_LivingArea
IQR_LivingArea
```

```python
#no living area outlier
df = df[(df.get('LivingArea') >= Q1_LivingArea - 1.5*IQR_LivingArea) & (df.get('LivingArea') <= Q3_LivingArea+1.5*IQR_LivingArea)]
df
```

```python
IQR_ParkingTotal = Q3_ParkingTotal - Q1_ParkingTotal
IQR_ParkingTotal
```
