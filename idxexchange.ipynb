```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dec2024 = pd.read_csv("CRMLSSold202412.csv")
jan2025 = pd.read_csv("CRMLSSold202501_filled.csv")
feb2025 = pd.read_csv("CRMLSSold202502.csv")
mar2025 = pd.read_csv("CRMLSSold202503.csv")
apr2025 = pd.read_csv("CRMLSSold202504.csv")
may2025 = pd.read_csv("CRMLSSold202505.csv")
result = pd.concat([dec2024, jan2025, feb2025, mar2025, apr2025,may2025], ignore_index=True)
display(result)
```

```python
result.columns
```

```python
result = result[(result["PropertyType"] == "Residential") & (result["PropertySubType"]== "SingleFamilyResidence")]
result
```

```python
result.columns
```

```python
result = result.drop(columns = ['BuyerAgentAOR', 'ListAgentAOR', 'Flooring','WaterfrontYN','OriginalListPrice', 'ListingKey', 
                                'ListAgentEmail', 'CloseDate', 'ClosePrice', 'ListAgentFirstName','ListAgentLastName',
                                'UnparsedAddress', 'PropertyType','ListPrice', 'DaysOnMarket', 'ListOfficeName', 
                                'BuyerOfficeName', 'CoListOfficeName', 'ListAgentFullName', 'CoListAgentFirstName', 
                                'CoListAgentLastName', 'BuyerAgentMlsId', 'BuyerAgentFirstName', 'BuyerAgentLastName',
                                'AssociationFeeFrequency', 'AboveGradeFinishedArea', 'ListingKeyNumeric', 'MLSAreaMajor', 
                                'TaxAnnualAmount','MlsStatus', 'ElementarySchool', 'BuilderName', 'PropertySubType', 
                                'SubdivisionName', 'BuyerOfficeAOR', 'ListingId','TaxYear', 'BuildingAreaTotal',
                                'ContractStatusChangeDate', 'ElementarySchoolDistrict', 'CoBuyerAgentFirstName', 
                                'PurchaseContractDate', 'ListingContractDate', 'BelowGradeFinishedArea', 'BusinessType',
                                'CoveredSpaces', 'MiddleOrJuniorSchool','HighSchool','LotSizeDimensions','MainLevelBedrooms',
                                'AssociationFee','MiddleOrJuniorSchoolDistrict', 'latfilled', 'lonfilled'])
result
```

```python
# Keep rows within California bounds
result = result[((result["Longitude"] >= -124.409591) & (result["Longitude"] <= -114.131211)) & ((result["Latitude"] >= 32.534156) & (result["Latitude"] <= 42.009518))]
```

```python
result.columns
```

```python
for var in ['LotSizeAcres', 'LotSizeSquareFeet','LotSizeArea']:
    print(var,": ", result[var].isna().mean())
```

```python
result = result[(result['YearBuilt'].notna()) & (result['YearBuilt'] <= 2025)]
# drop rows with NaN and 2026 for YearBuilt
result['BuildingAge'] = 2025 - result['YearBuilt']
# create a new age column
result = result[(result['BathroomsTotalInteger'].notna()) & (~result['BathroomsTotalInteger'].isin([25, 21, 20, 45, 153]))]
# drop bathroom numbers of 25, 21, 20, 45, 153 and nan
result = result.dropna(subset = ['City'])
```

```python
# 1. New Construction
# If NA means "not new construction" in context, replace NA with False. Cause if it is a new constuction, it should be actively posted online.
result['NewConstructionYN'] = result['NewConstructionYN'].fillna(False)
# 2. Garage Space: NA probably means 0 garage spaces. Due to the same reason above.
# e.g. condo, older house 
result['GarageSpaces'] = result['GarageSpaces'].fillna(0)
result['PostalCode'] = result['PostalCode'].fillna('Unknown')
col = 'GarageSpaces'
q1 = result[col].quantile(0.25)
q3 = result[col].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
result = result[result[col].isna() | ((result[col] >= lower) & (result[col] <= upper))]
# For general California housing analysis, keep only homes with LotSizeSquareFeet ≤ 20,000 sqft
result = result[result['LotSizeSquareFeet'].isna() | (result['LotSizeSquareFeet'] <= 20000)]
```

```python
# --- 1. Clean Boolean Columns ---
bool_cols = ['ViewYN', 'PoolPrivateYN', 'AttachedGarageYN', 'FireplaceYN']
for col in bool_cols:
    if col in result.columns:
        result[col] = result[col].astype(str).str.strip().str.lower().map({'true': True, 'false': False})
        result[col] = result[col].fillna(False)  # assuming missing = False
```

```python
result['StoriesFinal'] = result['Stories'].fillna(result['Levels'])
result = result.drop(columns=['Stories', 'Levels'])
```

```python
for var in ['ParkingTotal', 'GarageSpaces']:
    print(var,": ", result[var].isna().mean())
display(result['ParkingTotal'])
display(result['GarageSpaces'])
```

```python
# after EDA, columns to exclude 
columns_to_exclude = ['StreetNumberNumeric', 'LotSizeAcres', 'LotSizeSquareFeet','FireplacesTotal', 'BasementYN','StateOrProvince','YearBuilt', 'GarageSpaces']
result = result.drop(columns = columns_to_exclude)
```
